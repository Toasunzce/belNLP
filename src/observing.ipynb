```python
from belNLP.morphology.pos_tagger import POSTagger
from belNLP.morphology.lemmatizer import Lemmatizer
from belNLP.tokenization.tokenizers import WordTokenizer
import warnings
warnings.filterwarnings('ignore')
```

```python
tokenizer = WordTokenizer()
tagger = POSTagger.load('models/POSTagger.pt')
lemmatizer = Lemmatizer.load('models/Lemmatizer.pt')
```

```python
text = "Фасады гарызантальна ашаляваны, прарэзаны лучковымі аконнымі праёмамі."


tokens = tokenizer(text)
morph = tagger(tokens)
morph = lemmatizer(morph)


print(f"{"word":<14} | {"lemmatized":<14} | {"tag":<6} | {"probability"}")
print("="*53)
for token in morph:
    base_text = token.text
    lemmatized = token.lemma
    tag = token.pos
    if tag in ["X", "SYM"]:
        continue
    prob = float(token.morph['pos_prob'])  # ty:ignore[not-subscriptable]
    print(f"{base_text:<14} | {lemmatized:<14} | {tag:<6} | {prob:.5f}")

print('\n\n', morph[0])
```
